```python
#add 10 with all the element in the given tuple
tu = (3,4,5,6)
for i in tu:
    newvalue = i+10
    print(newvalue)
print('The elements in the list is completed')
```

```python
set1={55,21,78,13}
for r in set1:
    print(r)
```

```python
dic={1:'Kalpana',2:'Geetha', 3:'Pragalya'}
for i in dic:
    print(dic[i])
print('printed all the names')
```

```python
dic={1:'Kalpana',2:'Geetha', 3:'Pragalya'}
for i in dic.values():
    print(i)
print('printed all the names')
```

```python
#nested for loop
lis1 = ['red','black','green','pink','yellow']
lis2=[20,30,40]
for i in lis1:
    for k in lis2:
        print(i,k)
```
